# Starter Code

```python
import random  
import gym  
import numpy as np  
from collections import deque  
from keras.models import Sequential  
from keras.layers import Dense  
from keras.optimizers import Adam  
  
  
from scores.score_logger import ScoreLogger  
  
ENV_NAME = "CartPole-v1"  
  
GAMMA = 0.95  
LEARNING_RATE = 0.001  
  
MEMORY_SIZE = 1000000  
BATCH_SIZE = 20  
  
EXPLORATION_MAX = 1.0  
EXPLORATION_MIN = 0.01  
EXPLORATION_DECAY = 0.995  
  
  
class DQNSolver:  
  
    def __init__(self, observation_space, action_space):  
        self.exploration_rate = EXPLORATION_MAX  
  
        self.action_space = action_space  
        self.memory = deque(maxlen=MEMORY_SIZE)  
  
        self.model = Sequential()  
        self.model.add(Dense(24, input_shape=(observation_space,), activation="relu"))  
        self.model.add(Dense(24, activation="relu"))  
        self.model.add(Dense(self.action_space, activation="linear"))  
        self.model.compile(loss="mse", optimizer=Adam(lr=LEARNING_RATE))  
  
    def remember(self, state, action, reward, next_state, done):  
        self.memory.append((state, action, reward, next_state, done))  
  
    def act(self, state):  
        if np.random.rand() < self.exploration_rate:  
            return random.randrange(self.action_space)  
        q_values = self.model.predict(state)  
        return np.argmax(q_values[0])  
  
    def experience_replay(self):  
        if len(self.memory) < BATCH_SIZE:  
            return  
        batch = random.sample(self.memory, BATCH_SIZE)  
        for state, action, reward, state_next, terminal in batch:  
            q_update = reward  
            if not terminal:  
                q_update = (reward + GAMMA * np.amax(self.model.predict(state_next)[0]))  
            q_values = self.model.predict(state)  
            q_values[0][action] = q_update  
            self.model.fit(state, q_values, verbose=0)  
        self.exploration_rate *= EXPLORATION_DECAY  
        self.exploration_rate = max(EXPLORATION_MIN, self.exploration_rate)  
  
  
def cartpole():  
    env = gym.make(ENV_NAME)  
    score_logger = ScoreLogger(ENV_NAME)  
    observation_space = env.observation_space.shape[0]  
    action_space = env.action_space.n  
    dqn_solver = DQNSolver(observation_space, action_space)  
    run = 0  
    while True:  
        run += 1  
        state = env.reset()  
        state = np.reshape(state, [1, observation_space])  
        step = 0  
        while True:  
            step += 1  
            #env.render()  
            action = dqn_solver.act(state)  
            state_next, reward, terminal, info = env.step(action)  
            reward = reward if not terminal else -reward  
            state_next = np.reshape(state_next, [1, observation_space])  
            dqn_solver.remember(state, action, reward, state_next, terminal)  
            state = state_next  
            if terminal:  
                print ("Run: " + str(run) + ", exploration: " + str(dqn_solver.exploration_rate) + ", score: " + str(step))  
                score_logger.add_score(step, run)  
                break  
            dqn_solver.experience_replay()  

```

The goal of the agent within the cartpole problem is to keep a pole balanced upright in a cart for as long as possible. If the pole falls past a certain angle or the cart goes beyond the bounds of the track, the agent loses. The different states are the cart position, cart velocity, pole angle, and pole angular velocity. These values are represented within the program as a state vector. An agent within this simulation has two possible actions, push the cart to left or push the cart to the right. The reinforment learning algorithm being used is Deep Q-Learning (DQN). DQN is an improved version of Q-Learning. It enables agents to learn smarter and faster by leveraging a neural network instead of a large lookup table, balancing exploration and exploitation, remembering and recalling past experience, and using networks to avoid wild updates (Beysolow, 2019).

The purpose of experience replay is to improve training by letting the agent learn from past experience. In this algorithm, a replay buffer of size one million is leveraged to store experiences. Every time a step completes, the agent uses the `remember()` method to save to this memory. At a certain point, the `experience_replay()` method is called to recall a random batch of experiences. For this random batch of experiences, the Q-value is updated using the Bellman equation. The model is then trained to better it's Q-value prediction. A discount factor (`GAMMA`) is also introduced which determines how much the agent will favor short-term vs. long-term rewards. In the starter code, this is set to `0.95` which encourages the agent to choose sustained rewards over time while still accounting for short-term outcomes.

The artificial neural network used in this algorithm has an input layer, two hidden layers, and an output layer. It uses the Adam optimizer and the MSE loss function. The input layer takes in a state vector of cart position, cart velocity, pole angle, and pole angular velocity. The first and seccond hidden layers have twenty-four neurons and use the ReLU activation function. The output layer has two neurons, one for move left and one for move right, and uses a linear activation function. It outputs a Q-value which represents the expected future reward of each action at the current state. By leveraging a neural network, the algorithm becomes more efficient. Since the algorithm can approximate Q-values without using a lookup table, it does not have to memorize each state-action pair in a massive lookup table which would prove nigh impossible in environments with a large number of possible states and actions (Singh, 2022).
