### Q-Learning Function

Q[state, action] = R(state, action) + Gamma * Max[Q(next state, all actions)]

Q: A quality function that for a given state and action returns a value of that state-action pair.

R: A reward function that we use to reinforce agent behavior.

Gamma: A number between 0 and 1 that determines whether an agent will value imeediate reward from R or a delayed reward from Q.
